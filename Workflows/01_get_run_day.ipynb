```python
_input_date = dbutils.widgets.get("Input_date")
print(_input_date)
```

```python
spark.conf.set("spark.sql.legacy.timeParserPolicy", "LEGACY")

input_day = spark.sql(f"""
select  date_format(try_to_timestamp('{_input_date}',"yyyy-MM-dd'T'HH:mm:ss"),'E')
""").collect()[0][0]
print(input_day)
```

```python
dbutils.jobs.taskValues.set("input_day",input_day)
```
